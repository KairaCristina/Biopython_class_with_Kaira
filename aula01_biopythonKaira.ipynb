## Realizando algumas análises

```python
from Bio.Seq import Seq

# Exemplo de sequência de DNA
sequencia = Seq("ATGCTGATGC")
sequencia
```

```python
# contando um nucleotídeo específico
sequencia.count("a")
```

```python
from Bio.SeqUtils import gc_fraction

# Calcular o conteúdo de GC
gc = gc_fraction(sequencia)

# Calcular a fração de AT
at = 1 - gc

# Imprimir os resultados
print(f"Conteúdo de GC: {gc}")
print(f"Fração unitária de AT: {at}")
```

```python
reverso = sequencia.reverse_complement()
reverso
```

```python
# Traduzindo a sequência de DNA para mRNA
mRNA = sequencia.transcribe()
print(mRNA)
```

```python
# Traduzindo a sequência de DNA para proteína
proteina = sequencia.translate()
print(proteina)
```

```python
from Bio.Data import IUPACData

# Obtendo os códigos de três letras a partir de uma letra
codigos_tres_letras = [IUPACData.protein_letters_1to3[aa] for aa in proteina]
print(codigos_tres_letras)
```

```python
from Bio import Align

# Criar um alinhamento de sequências
aligner = Align.PairwiseAligner()
seq1 = "ACTGCTACTGCATGCA"
seq2 = "AGGAGGCTTCAGTCTT"
alignment = aligner.align(seq1, seq2)

# Imprimir o alinhamento e score
for res in alignment:
    print(res)
    print(res.score)
```

# Agora vamos desvendar um caso clínico!

Um homem de 20 anos procurou um serviço de saúde, relatando um histórico de:
* dois dias de febre,
* calafrios,
* mialgia (dores musculares),
* faringite (dor de garganta)
* e dor de cabeça.

A doença foi de curta duração; sua temperatura voltou ao normal após três dias, mas ele continuou com mialgia persistente e se sentiu cansado por uma semana. Ele se recuperou completamente.

Ao fazer o sequenciamento de amostra do paciente, foi encontrado em um patógeno desconhecido cuja sequencia nucleotídea está no arquivo 'patogeno1.fasta'.

Vamos desvendar:
 * qual é esse patógeno
 * qual o organismo
 * onde ele já apareceu (região)

```python
from Bio.Blast import NCBIWWW
from Bio import SeqIO

# Carrega a sequência do arquivo FASTA
file = open("patogeno1.fasta", 'r')
for record in SeqIO.parse(file, 'fasta'):
    sequencia = record.seq
```

```python
sequencia.count("GC")
```

```python
# Calcular o conteúdo de GC
gc = gc_fraction(sequencia)

# Calcular a fração de AT
at = 1 - gc

# Imprimir os resultados
print(f"Conteúdo de GC: {gc}")
print(f"Fração unitária de AT: {at}")
```

```python
from Bio.Blast import NCBIXML

# Abre o arquivo XML com os resultados do BLAST
with open("resultado_blast.xml", "r") as arquivo_blast:
    resultados = NCBIXML.parse(arquivo_blast)

    # Analisa os resultados
    for resultado in resultados:
        for alinhamento in resultado.alignments:
            print(f"\nOrganismo: {alinhamento.hit_def}")
            print(f"ID de acesso: {alinhamento.accession}")
            print(f"Score: {alinhamento.hsps[0].score}")
            print(f"E-value: {alinhamento.hsps[0].expect}")
            print(f"Alinhamento: {alinhamento.hsps[0].query}")
            print(f"             {alinhamento.hsps[0].match}")
            print(f"             {alinhamento.hsps[0].sbjct}")
```

### Interpretação dos resultados:

* O E-value indica a significância estatística do alinhamento. Quanto menor o E-value, mais significativo é o resultado.

* O Score indica a qualidade do alinhamento. Quanto maior o score, melhor o alinhamento.

```python
from Bio import Entrez
from Bio import SeqIO

# Configuração do email para acessar o NCBI (obrigatório)
Entrez.email = "k192007@dac.unicamp.br"  # Substitua pelo seu email

# Função para buscar informações de localização no GenBank
def buscar_regioes_patogeno(id_acesso):
    try:
        # Busca o registro no GenBank usando o ID de acesso
        handle = Entrez.efetch(db="nucleotide", id=id_acesso, rettype="gb", retmode="text")
        registro_genbank = SeqIO.read(handle, "genbank")  # Usando SeqIO para ler o registro
        handle.close()

        # Analisa o registro para encontrar informações de localização
        regioes = []
        for feature in registro_genbank.features:
            if feature.type == "source":  # Procura pela feature "source"
                if "country" in feature.qualifiers:  # Verifica se há informação de país
                    regiao = feature.qualifiers["country"][0]  # Extrai o nome da região
                    regioes.append(regiao)
                elif "geo_loc_name" in feature.qualifiers:  # Verifica o campo geo_loc_name
                    regiao = feature.qualifiers["geo_loc_name"][0]  # Extrai o nome da região
                    regioes.append(regiao)
                else:
                    print("Nenhum campo de localização encontrado na feature 'source'.")

        return regioes
    except Exception as e:
        print(f"Erro ao buscar no GenBank: {e}")
        return []
```

```python
# Exemplo de uso
id_acesso = "NC_009029"  # Substitua pelo ID de acesso do patógeno
regioes = buscar_regioes_patogeno(id_acesso)

if regioes:
    print(f"O patógeno foi encontrado nas seguintes regiões: {', '.join(regioes)}")
else:
    print("Nenhuma informação de localização encontrada.")
```
